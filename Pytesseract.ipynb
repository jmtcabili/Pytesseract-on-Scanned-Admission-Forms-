```python
import pytesseract
from PIL import Image

# Specify the path to the Tesseract executable
pytesseract.pytesseract.tesseract_cmd = r'C:\Program Files\Tesseract-OCR\tesseract.exe'
```

```python
# Open an image file
png_02021e = Image.open('img1.png')
png_1722e = Image.open('img2.png')
png_3224_67e = Image.open('img2.png')
png_on00600e_annexd = Image.open('img2.png')
```

```python
bbox_02021e = (0, 767, png_02021e.size[0], 1280) # Bounding Box for 02021e
bbox_1722e = (0, 862, png_1722e.size[0], 1120) # Bounding Box for 1722e
bbox_3224_67e = (0, 490, png_3224_67e.size[0], 980) # Bounding Box for 3224-67e
bbox_on00600e_annexd = (0, 800, png_on00600e_annexd.size[0], 1160) # Bounding Box for on00600e_annexd
```

```python
display(png_02021e.crop(bbox_1722e))
```
